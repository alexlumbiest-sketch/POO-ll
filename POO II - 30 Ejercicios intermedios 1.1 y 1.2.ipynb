Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:

```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
producto = input ("Ingrese el nombre del producto: ")
cantidad = int(input("Ingrese la cantidad: "))
precio_unitario = float(input("Ingrese el precio unitario: "))

subtotal = cantidad * precio_unitario
descuento = subtotal * 0.10  # 10% de descuento
iva = (subtotal - descuento) * 0.15
total = subtotal - descuento + iva

print(f"producto: {producto}")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Descuento: ${descuento:.2f}")
print(f"IVA: ${iva:.2f}")
print(f"Total a pagar: ${total:.2f}")


````

### Ejercicio 22 · Inventario con diccionario ★★

Con `stock = {"laptop": 12, "mouse": 45, "teclado": 30, "monitor": 8}`:
1. Se vendieron 5 mouses: actualice el stock.

3. Se dejó de vender el monitor: elimínelo con `pop` y muestre cuántas unidades tenía.
4. Consulte `"impresora"` con `get` para que muestre `0` si no existe.

### Ejercicio 24 · Operaciones con conjuntos ★★

Estudiantes inscritos en tres talleres:
```python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t  = {"Luis", "Carla", "Pedro", "Diego"}
ia_t     = {"Ana", "Pedro", "Diego"}
```
Muestre (ordenado con `sorted`):
1. Quiénes están en Python **y** en Excel.
2. Quiénes están en los tres talleres.
3. Quiénes están en Python pero **no** en IA.
4. Quiénes están en Python **o** en Excel, pero no en ambos (`^`).
5. Si todos los de IA están también en Python (`issubset`).
6. El total de estudiantes distintos.

````python
python_t = {"Ana", "Luis", "María", "Pedro", "Sofía"}
excel_t = {"Luis", "Carla", "Pedro", "Diego"}
ia_t = {"Ana", "Pedro", "Diego"}
# Escriba su solución aquí
print("Python y Excel:", sorted(python_t & excel_t))
print("Los tres talleres:", sorted(python_t & excel_t & ia_t))
print("Python pero no IA:", sorted(python_t - ia_t))
print("Python o Excel, no ambos:", sorted(python_t ^ excel_t))
print("¿IA dentro de Python?", ia_t.issubset(python_t))
print("Total distintos:", len(python_t | excel_t | ia_t))
````

### Ejercicio 27 · Revisar un correo electrónico ★★

Con `correo = "  Ana.Perez@TecAzuay.edu.EC "`:
1. Límpielo y páselo a minúsculas.
2. Separe el usuario y el dominio con `split("@")`.
3. Muestre `True` o `False` para cada condición (todavía sin `if`): tiene exactamente una `@`, el dominio termina en `".ec"`, el usuario tiene al menos 3 caracteres y no contiene espacios.
4. Muestre si el correo es válido: todas las condiciones a la vez, unidas con `and`.

````python
correo = "  Ana.Perez@TecAzuay.edu.EC "
# Escriba su solución aquí

correo = correo.strip().lower()
usuario, dominio = correo.split("@")

una_arroba = correo.count("@") == 1
termina_ec = dominio.endswith(".ec")
usuario_largo = len(usuario) >= 3
sin_espacios = " " not in correo

print("Una sola @:", una_arroba)
print("Dominio termina en .ec:", termina_ec)
print("Usuario con 3+ caracteres:", usuario_largo)
print("Sin espacios:", sin_espacios)
print("¿Correo válido?", una_arroba and termina_ec and usuario_largo and sin_espacios)
````

### Ejercicio 29 · Copia o referencia ★★

Ejecute y explique en comentarios cada resultado:
```python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()
alias["saldo"] = 500
copia["saldo"] = 999
```
1. Muestre `original`, `alias` y `copia`.
2. Compruebe con `is` cuáles son el mismo objeto.
3. Compare con `id()` las direcciones de memoria.

````python
original = {"nombre": "Ana", "saldo": 100}
alias = original
copia = original.copy()
alias["saldo"] = 500
copia["saldo"] = 999

print("original:", original)  
print("alias:   ", alias)    
print("copia:   ", copia)    

print("original is alias:", original is alias)   # True
print("original is copia:", original is copia)   # False

print("id original:", id(original))
print("id alias:   ", id(alias))   
print("id copia:   ", id(copia))    
````

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"

partes = linea.split("|")
nombre = partes[0].title()
cedula = partes[1]

tipo1, saldo1 = partes[2].split("=")
tipo2, saldo2 = partes[3].split("=")
cuentas = {tipo1: float(saldo1), tipo2: float(saldo2)}

socio = {"nombre": nombre, "cedula": cedula, "cuentas": cuentas}

print(f"Socio: {socio['nombre']} ({socio['cedula']})")
print(f"Tipos de cuenta: {', '.join(socio['cuentas'])}")
print(f"Saldo total: ${sum(socio['cuentas'].values()):,.2f}")
````
